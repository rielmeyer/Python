- количество строк;
- количество слов;
- количество символов;
- длину самой длинной строки.

> **Примечание:** При обработке файла не загружайте всё содержимое целиком в память.

1. Загрузите и откройте файл txt с текстом.

````python
from pathlib import Path

print(Path(r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\text.txt").read_text(encoding="utf-8"))
````

2. Превратите файл в массив массивов: каждый подмассив - массив символов одной строки из текста

Назовите его result (для дальнейших решений)

> **Подсказка 1:** .splitlines() - разбивает одну большую строку на список отдельных строк, разделяя её по символам переноса строки

> **Подсказка 2:** .line() - превращает строку в массив символов

````python
path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\text.txt"

# Создаём пустой список для итогового результата
result = []

# Открываем файл в режиме чтения
with open(path, "r", encoding="utf-8") as file:
    # Читаем текст и разбиваем на отдельные строки
    lines = file.read().splitlines()
    
    # Проходим по каждой полученной строке
    for line in lines:
        # Превращаем строку в массив символов и добавляем в общий массив
        result.append(list(line))

# Проверяем структуру (например, выводим первую строку как массив символов)
print(result)
````

4. Найдите количество слов

> **Подсказка 1:** Посчитайте количество символов для каждой строки.

> **Подсказка 2:** a.isspace() - проверяет, является ли символ a пробелом

> **Подсказка 3:** В каждой строке идите последовательно по символам. Заведите счетчик. Как только он будет видеть пробел или переход на следующую строку - он будет выключаться. Когда увидит символ - включится. По количеству включений счетчика получится число слов.

````python
# Создаем счетчик общего количества слов в тексте
word_count = 0

# Последовательно перебираем все строки из списка result
for line in result:

    # Считаем количество символов в текущей строке согласно подсказке 1
    line_length = len(line)

    # Создаем счетчик позиции символа в текущей строке
    i = 0

    # Создаем переключатель, который показывает, находимся ли мы сейчас внутри слова
    in_word = False

    # Последовательно проходим по всем символам текущей строки
    while i < line_length:

        # Получаем текущий символ строки
        symbol = line[i]

        # Проверяем, является ли текущий символ пробельным
        if symbol.isspace():

            # Если встретили пробел, считаем, что сейчас мы не внутри слова
            in_word = False

        # Если символ не является пробельным
        else:

            # Проверяем, был ли счетчик до этого выключен
            if not in_word:

                # Увеличиваем количество слов, так как началось новое слово
                word_count += 1

                # Включаем счетчик: теперь мы находимся внутри слова
                in_word = True

        # Переходим к следующему символу строки
        i += 1

# Выводим найденное количество слов
print("Количество слов:", word_count)
````

5. Найдите количество символов

> **Подсказка 1:** a.isspace() - проверяет, является ли символ a пробелом

> **Подсказка 2:** Идите последовательно по каждой строке массива result и смотрите является ли символ пробелом. Если не является - повысьте счетчик на 1

````python
# Создаем счетчик количества символов, не являющихся пробельными
symbol_count = 0

# Последовательно перебираем каждую строку из списка result
for line in result:

    # Последовательно перебираем каждый символ текущей строки
    for symbol in line:

        # Проверяем, что текущий символ не является пробельным
        if not symbol.isspace():

            # Если символ не пробельный, увеличиваем счетчик на 1
            symbol_count += 1

# Выводим количество символов без учета пробельных символов
print("Количество символов:", symbol_count)
````

**Подсказка 2:** Решайте через цикл, постепенно обнавляя максимальную длину.

### 3. Обработка CSV

Файл `students.csv` имеет вид:

```csv
name,group,score
Alice,101,87
Bob,102,91
Charlie,101,73
David,102,95
Eva,101,88
```

Напишите программу, которая определяет:
- средний балл всех студентов;
- студента с максимальным баллом;
- средний балл каждой группы.

> **Примечание:** Используйте модуль `csv`.

````python
# Импортируем модуль csv
import csv

# Создаем пустой список для хранения данных о студентах
students = []

# Открываем файл students.csv для чтения
with open(r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\students.csv",
          "r", encoding="utf-8") as file:

    # Читаем CSV-файл с учетом названий столбцов
    reader = csv.DictReader(file)

    # Последовательно перебираем строки файла
    for row in reader:

        # Преобразуем балл студента из строки в целое число
        row["score"] = int(row["score"])

        # Добавляем данные студента в список
        students.append(row)


# Создаем переменную для хранения суммы баллов всех студентов
total_score = 0

# Последовательно перебираем всех студентов
for student in students:

    # Прибавляем балл текущего студента к общей сумме
    total_score += student["score"]

# Вычисляем средний балл всех студентов
average_score = total_score / len(students)

# Выводим средний балл всех студентов
print("Средний балл всех студентов:", average_score)


# Считаем первого студента студентом с максимальным баллом
best_student = students[0]

# Последовательно перебираем всех студентов
for student in students:

    # Проверяем, больше ли балл текущего студента максимального найденного балла
    if student["score"] > best_student["score"]:

        # Обновляем студента с максимальным баллом
        best_student = student

# Выводим имя студента с максимальным баллом
print("Студент с максимальным баллом:", best_student["name"])

# Выводим его балл
print("Максимальный балл:", best_student["score"])


# Создаем словарь для хранения суммы баллов каждой группы
group_sum = {}

# Создаем словарь для хранения количества студентов каждой группы
group_count = {}

# Последовательно перебираем всех студентов
for student in students:

    # Получаем номер группы текущего студента
    group = student["group"]

    # Получаем балл текущего студента
    score = student["score"]

    # Проверяем, встречалась ли эта группа раньше
    if group not in group_sum:

        # Создаем для новой группы начальную сумму баллов
        group_sum[group] = 0

        # Создаем для новой группы счетчик студентов
        group_count[group] = 0

    # Добавляем балл студента к сумме баллов его группы
    group_sum[group] += score

    # Увеличиваем количество студентов этой группы на 1
    group_count[group] += 1


# Последовательно перебираем все группы
for group in group_sum:

    # Вычисляем средний балл текущей группы
    group_average = group_sum[group] / group_count[group]

    # Выводим номер группы и ее средний балл
    print("Средний балл группы", group, ":", group_average)
````

### 4. Фильтрация CSV

Дан файл с данными о товарах: `expensive_.csv`

Создайте новый файл `expensive.csv`, содержащий только товары стоимостью более 100.

> **Примечание:** Сохраните исходный заголовок CSV.

````python
# Импортируем модуль csv
import csv

# Указываем путь к исходному файлу
input_path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\expensive_.csv"

# Указываем путь к новому файлу
output_path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\expensive.csv"

# Открываем исходный файл для чтения
with open(input_path, "r", encoding="utf-8") as input_file:

    # Читаем CSV-файл с учетом заголовков
    reader = csv.DictReader(input_file)

    # Открываем новый файл для записи
    with open(output_path, "w", newline="", encoding="utf-8") as output_file:

        # Создаем объект для записи с теми же названиями столбцов
        writer = csv.DictWriter(output_file, fieldnames=reader.fieldnames)

        # Записываем исходный заголовок CSV в новый файл
        writer.writeheader()

        # Последовательно перебираем все товары
        for row in reader:

            # Преобразуем стоимость товара в число и проверяем, больше ли она 100
            if float(row["price"]) > 100:

                # Записываем подходящий товар в новый файл
                writer.writerow(row)
````

### 5. JSON и преобразование данных

Дан файл `students.json`.

Напишите программу, которая:
- загружает данные из JSON;
- добавляет каждому студенту поле `"passed"`;
- определяет студента с максимальным баллом;
- сохраняет изменённые данные в новый JSON-файл.

> **Примечание:** Студент считается успешно сдавшим экзамен, если его балл не меньше 60.

````python
# Импортируем модуль json
import json

# Указываем путь к исходному JSON-файлу
input_path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\students.json"

# Указываем путь к новому JSON-файлу
output_path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\students_result.json"

# Открываем исходный JSON-файл для чтения
with open(input_path, "r", encoding="utf-8") as file:

    # Загружаем данные из JSON-файла
    students = json.load(file)


# Последовательно перебираем всех студентов
for student in students:

    # Проверяем, набрал ли студент не меньше 60 баллов
    if student["score"] >= 60:

        # Добавляем поле passed со значением True
        student["passed"] = True

    # Если студент набрал меньше 60 баллов
    else:

        # Добавляем поле passed со значением False
        student["passed"] = False


# Считаем первого студента студентом с максимальным баллом
best_student = students[0]

# Последовательно перебираем всех студентов
for student in students:

    # Проверяем, больше ли балл текущего студента максимального найденного балла
    if student["score"] > best_student["score"]:

        # Обновляем студента с максимальным баллом
        best_student = student


# Выводим студента с максимальным баллом
print("Студент с максимальным баллом:", best_student["name"])

# Выводим его балл
print("Максимальный балл:", best_student["score"])


# Открываем новый JSON-файл для записи
with open(output_path, "w", encoding="utf-8") as file:

    # Сохраняем измененные данные в новый JSON-файл
    json.dump(students, file, ensure_ascii=False, indent=4)
````

### 6. Поиск телефонных номеров

Дана файл phone_numbers.txt, содержащий произвольный текст. Найдите в ней все телефонные номера в формате:
* `+7 999 123-45-67`
* `8 999 123-45-67`
* `+7 (999) 123-45-67`
* `8 (999) 123-45-67`

Выведите найденные номера.

````python
# Указываем путь к файлу
file_path = r"C:\Users\koval\OneDrive\Desktop\МИФИ\Python\4. files\phone_numbers.txt"

# Открываем файл для чтения
with open(file_path, "r", encoding="utf-8") as file:

    # Считываем весь текст из файла
    text = file.read()


# Создаем пустой список для найденных телефонных номеров
phone_numbers = []

# Начинаем просмотр текста с первого символа
i = 0

# Последовательно перебираем символы текста
while i < len(text):

    # Создаем пустую переменную для найденного номера
    number = None

    # Проверяем формат +7 999 123-45-67
    if text[i:i + 16].startswith("+7 "):

        # Берем 16 символов, начиная с текущей позиции
        candidate = text[i:i + 16]

        # Проверяем все части номера
        if (len(candidate) == 16
                and candidate[3:6].isdigit()
                and candidate[6] == " "
                and candidate[7:10].isdigit()
                and candidate[10] == "-"
                and candidate[11:13].isdigit()
                and candidate[13] == "-"
                and candidate[14:16].isdigit()):

            # Запоминаем найденный номер
            number = candidate


    # Проверяем формат 8 999 123-45-67
    if number is None and text[i:i + 15].startswith("8 "):

        # Берем 15 символов, начиная с текущей позиции
        candidate = text[i:i + 15]

        # Проверяем все части номера
        if (len(candidate) == 15
                and candidate[2:5].isdigit()
                and candidate[5] == " "
                and candidate[6:9].isdigit()
                and candidate[9] == "-"
                and candidate[10:12].isdigit()
                and candidate[12] == "-"
                and candidate[13:15].isdigit()):

            # Запоминаем найденный номер
            number = candidate


    # Проверяем формат +7 (999) 123-45-67
    if number is None and text[i:i + 18].startswith("+7 ("):

        # Берем 18 символов, начиная с текущей позиции
        candidate = text[i:i + 18]

        # Проверяем все части номера
        if (len(candidate) == 18
                and candidate[4:7].isdigit()
                and candidate[7] == ")"
                and candidate[8] == " "
                and candidate[9:12].isdigit()
                and candidate[12] == "-"
                and candidate[13:15].isdigit()
                and candidate[15] == "-"
                and candidate[16:18].isdigit()):

            # Запоминаем найденный номер
            number = candidate


    # Проверяем формат 8 (999) 123-45-67
    if number is None and text[i:i + 17].startswith("8 ("):

        # Берем 17 символов, начиная с текущей позиции
        candidate = text[i:i + 17]

        # Проверяем все части номера
        if (len(candidate) == 17
                and candidate[3:6].isdigit()
                and candidate[6] == ")"
                and candidate[7] == " "
                and candidate[8:11].isdigit()
                and candidate[11] == "-"
                and candidate[12:14].isdigit()
                and candidate[14] == "-"
                and candidate[15:17].isdigit()):

            # Запоминаем найденный номер
            number = candidate


    # Проверяем, нашли ли мы телефонный номер
    if number is not None:

        # Добавляем найденный номер в список
        phone_numbers.append(number)

        # Переходим сразу за найденный телефонный номер
        i += len(number)

    # Если телефонный номер не найден
    else:

        # Переходим к следующему символу текста
        i += 1


# Последовательно выводим все найденные телефонные номера
for number in phone_numbers:

    # Выводим номер
    print(number)
````

* начинается с латинской буквы;
* содержит только латинские буквы, цифры и символ _.

````python
def is_valid_login(login):

    # Проверяем длину логина
    if len(login) < 4 or len(login) > 16:
        return False

    # Проверяем, что первый символ является латинской буквой
    if not (('a' <= login[0] <= 'z') or ('A' <= login[0] <= 'Z')):
        return False

    # Последовательно перебираем каждый символ логина
    for symbol in login:

        # Проверяем, является ли символ строчной латинской буквой
        is_lower_letter = 'a' <= symbol <= 'z'

        # Проверяем, является ли символ заглавной латинской буквой
        is_upper_letter = 'A' <= symbol <= 'Z'

        # Проверяем, является ли символ цифрой
        is_digit = '0' <= symbol <= '9'

        # Проверяем, является ли символ знаком подчеркивания
        is_underscore = symbol == '_'

        # Если символ не относится ни к одному разрешенному типу
        if not (is_lower_letter or is_upper_letter or is_digit or is_underscore):

            # Логин некорректный
            return False

    # Если все проверки пройдены, логин корректный
    return True
````

Тест

````python
print(is_valid_login("Alice_123"))   # True
print(is_valid_login("123Alice"))    # False
print(is_valid_login("Al"))          # False
print(is_valid_login("Alice-123"))   # False
````

### 8. Замена лишних пробелов

Дана строка:

```python
text = "Это   строка    с     большим    количеством   пробелов."
```

С помощью регулярного выражения замените последовательности из нескольких пробелов одним пробелом.

Результат:

```text
"Это строка с большим количеством пробелов."
```

Дополнительно обработайте строку:

```python
"   Текст   с   пробелами   "
```

так, чтобы в начале и конце строки пробелов не осталось.

````python
# Исходная строка
text = "   Текст   с   пробелами   "

# Создаем пустую строку для результата
result = ""

# Запоминаем, был ли предыдущий символ пробелом
previous_space = False

# Последовательно перебираем все символы строки
for symbol in text:

    # Проверяем, является ли символ пробелом
    if symbol == " ":

        # Добавляем пробел, только если результат уже не пустой
        # и предыдущий символ не был пробелом
        if result != "" and not previous_space:
            result += " "

        # Запоминаем, что встретили пробел
        previous_space = True

    # Если символ не является пробелом
    else:

        # Добавляем символ в результат
        result += symbol

        # Запоминаем, что текущий символ не является пробелом
        previous_space = False

# Проверяем, остался ли пробел в конце строки
if len(result) > 0 and result[-1] == " ":

    # Удаляем последний пробел
    result = result[:-1]

# Выводим результат
print(result)
````
